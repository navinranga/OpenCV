##### Basic functions

grey

```python
import cv2
import numpy as np

img = cv2.imread('resources/lena.png')
#kernel definition
kernel = np.ones((5,5), np.uint8)

imgGray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
#kSize : kernel size (should be odd number) and sigmaX : standard deviation in X direction
imgBlur = cv2.GaussianBlur(imgGray, (7,7), 0)
#Canny edge detector with threshold values, to reduce edges, increase threshold
imgCanny = cv2.Canny(img, 150, 150)
# dialation to increase the size of the edges with kernel
# the more the iterations, the more the edges
imgDialation = cv2.dilate(imgCanny, (7,7), iterations=2)
# erosion to decrease the size of the edges, make it thinner
imgEroded = cv2.erode(imgDialation, (7,7), iterations=2)

cv2.imshow('Image Gray', imgGray)
cv2.imshow('Image Blur', imgBlur)
cv2.imshow('Image Canny', imgCanny)
cv2.imshow('Image Dialation', imgDialation)
cv2.imshow('Image Eroded', imgEroded)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

##### Warp perspective

```python
import cv2
import numpy as np
img = cv2.imread('resources/joker.jpg')
width, height = 250, 350
pointsOriginal = np.float32([[1212, 338], [2612, 709], [1769, 2507], [270, 2038]])
pointsTransformed = np.float32([[0, 0], [width, 0], [width, height], [0, height]])

matrix = cv2.getPerspectiveTransform(pointsOriginal, pointsTransformed)
imgOutput = cv2.warpPerspective(img, matrix, (width, height))

#scale the image to fit the screen
img = cv2.resize(img, (0, 0), fx=0.1, fy=0.1)
cv2.imshow('Image original', img)
cv2.imshow('Image output', imgOutput)
cv2.waitKey(0)
cv2.destroyAllWindows()

# tips
# 1. pointsOriginal and pointsTransformed should be in same order
```

stack image function

```python
# stack images (detailed explanation ):
# https://www.geeksforgeeks.org/stacking-images-in-python-using-numpy/
def stackImages(scale,imgArray):
    rows = len(imgArray)
    cols = len(imgArray[0])
    rowsAvailable = isinstance(imgArray[0], list)
    width = imgArray[0][0].shape[1]
    height = imgArray[0][0].shape[0]
    if rowsAvailable :
        for x in range(0,rows):
            for y in range (0,cols):
                if imgArray[x][y].shape[:2] == imgArray[0][0].shape[:2]:
                    imgArray[x][y] = cv2.resize(imgArray[x][y], (0,0), None, scale, scale)
                else:
                    imgArray[x][y] = cv2.resize(imgArray[x][y], (imgArray[0][0].shape[1], imgArray[0][0].shape[0]), None, scale, scale)
                if len(imgArray[x][y].shape) == 2: imgArray[x][y]= cv2.cvtColor(imgArray[x][y], cv2.COLOR_GRAY2BGR)
        imageBlank = np.zeros((height, width, 3), np.uint8)
        hor = [imageBlank]*rows
        hor_con = [imageBlank]*rows
        for x in range(0,rows):
            hor[x] = np.hstack(imgArray[x])
        ver = np.vstack(hor)
    else:
        for x in range (0,rows):
            if imgArray[x].shape[:2] == imgArray[0].shape[:2]:
                imgArray[x] = cv2.resize(imgArray[x], (0,0), None, scale, scale)
            else:
                imgArray[x] = cv2.resize(imgArray[x], (imgArray[0].shape[1], imgArray[0].shape[0]), None, scale, scale)
            if len(imgArray[x].shape) == 2: imgArray[x] = cv2.cvtColor(imgArray[x], cv2.COLOR_GRAY2BGR)
        hor= np.hstack(imgArray)
        ver = hor
    return ver

imgStack = stackImages(0.25, ([img, img, img], [img, img, img], [img, img, img]))
cv2.imshow('Stacked Images', imgStack)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

##### Color detection

```python
import cv2
import numpy as np

# Path to the image file
path = "resources/lambo.png"

# Dummy callback function for trackbars (required by OpenCV)
def empty(a):
    pass

# Create a window named 'Trackbars' for the trackbars
cv2.namedWindow('Trackbars')
cv2.resizeWindow('Trackbars', 640, 240)  # Resize window for better visibility

# Create trackbars for adjusting HSV values
# Hue ranges from 0 to 179 in OpenCV
cv2.createTrackbar('Hue Min', 'Trackbars', 0, 179, empty)
cv2.createTrackbar('Hue Max', 'Trackbars', 179, 179, empty)
# Saturation ranges from 0 to 255
cv2.createTrackbar('Sat Min', 'Trackbars', 0, 255, empty)
cv2.createTrackbar('Sat Max', 'Trackbars', 255, 255, empty)
# Value (brightness) ranges from 0 to 255
cv2.createTrackbar('Val Min', 'Trackbars', 134, 255, empty)
cv2.createTrackbar('Val Max', 'Trackbars', 255, 255, empty)

# Read the image from disk
img = cv2.imread(path)
# Convert the image from BGR (OpenCV default) to HSV color space
imgHSV = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Start an infinite loop to update mask in real-time as trackbars are moved
while True:
    # Read the image again (to reset any changes)
    img = cv2.imread(path)
    # Convert to HSV color space
    imgHSV = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    
    # Get current positions of all trackbars
    h_min = cv2.getTrackbarPos('Hue Min', 'Trackbars')
    h_max = cv2.getTrackbarPos('Hue Max', 'Trackbars')
    s_min = cv2.getTrackbarPos('Sat Min', 'Trackbars')
    s_max = cv2.getTrackbarPos('Sat Max', 'Trackbars')
    v_min = cv2.getTrackbarPos('Val Min', 'Trackbars')
    v_max = cv2.getTrackbarPos('Val Max', 'Trackbars')
    
    # Print current HSV range for reference
    print(h_min, h_max, s_min, s_max, v_min, v_max)
    
    # Create numpy arrays for lower and upper HSV bounds
    lower = np.array([h_min, s_min, v_min])
    upper = np.array([h_max, s_max, v_max])
    
    # Create a mask: pixels within the HSV range are white (255), others are black (0)
    mask = cv2.inRange(imgHSV, lower, upper)
    # The mask highlights the region of interest in white
    
    # Apply the mask to the original image using bitwise AND
    # Only the selected color region will be visible, rest will be black
    imgResult = cv2.bitwise_and(img, img, mask=mask)
    
    # Stack images for visualization: original, HSV, mask, and result
    # stackImages is a custom function defined earlier in your notebook
    imgStack = stackImages(0.6, ([img, imgHSV], [mask, imgResult]))
    
    # Show the stacked images in a window
    cv2.imshow('stacked output', imgStack)
    
    # Wait for 1 ms and break the loop if 'q' is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Destroy all OpenCV windows after exiting the loop
cv2.destroyAllWindows()
```

##### Contours and shapes

```python
import cv2

# Function to find and draw contours on the image
def getContours(img):
    # Find contours in the image
    # cv2.RETR_EXTERNAL: retrieves only the extreme outer contours
    # cv2.CHAIN_APPROX_NONE: stores all the contour points (no compression)
    contours, hierarchy = cv2.findContours(img, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
    
    # Loop through each contour found
    for cnt in contours:
        # Calculate the area of the contour
        area = cv2.contourArea(cnt)
        print("area of contour ", area)  # Print the area for reference
        
        # Only process contours with area greater than 500 to filter out small/noisy contours
        if area > 500:
            # Draw the contour on imgContour
            # -1 means draw all contours, (255, 0, 0) is blue color, thickness=3
            cv2.drawContours(imgContour, cnt, -1, (255, 0, 0), 3)
            
            # Calculate the perimeter (arc length) of the contour
            # True means the contour is closed
            peri = cv2.arcLength(cnt, True)
            print("perimeter of contour:", peri)
            
            # Approximate the contour to get corner points (polygonal curve)
            # 0.02*peri: approximation accuracy (2% of perimeter)
            approx = cv2.approxPolyDP(cnt, 0.02*peri, True)
            print("number of corner points:", len(approx), "; points ", approx)
            
            # Get the number of corner points
            objCor = len(approx)
            
            # Get bounding rectangle coordinates around the approximated contour
            x, y, w, h = cv2.boundingRect(approx)
            
            # Identify the shape based on the number of corners
            if objCor == 3:
                objectType = "Tri"  # Triangle
            elif objCor == 4:
                # Calculate aspect ratio to distinguish square vs rectangle
                aspRatio = w / float(h)
                if aspRatio > 0.98 and aspRatio < 1.03:
                    objectType = "Square"
                else:
                    objectType = "Rectangle"
            elif objCor == 6:
                objectType = "Hexagon"
            elif objCor > 6:
                objectType = "Circle"
            else:
                objectType = "None"
            
            # Draw the bounding rectangle in red with thickness 5
            cv2.rectangle(imgContour, (x, y), (x + w, y + h), (0, 0, 255), 5)
            
            # Put text showing shape type and number of points near the rectangle
            cv2.putText(imgContour, objectType + ",Points: " + str(len(approx)), 
                        (x + w // 3 - 10, y + h // 3 - 10), cv2.FONT_HERSHEY_COMPLEX, .7,
                        (0, 0, 0), 2)
            # Optionally, put text showing area near the rectangle
            # cv2.putText(imgContour, "Area: " + str(int(area)), (x + w + 20, y + 45), cv2.FONT_HERSHEY_COMPLEX, .7, (0, 255, 0), 2)

# Path to the image containing shapes
path = "resources/shapes.png"

# Read the image
img = cv2.imread(path)

# Convert the image to grayscale (required for edge detection)
imgGray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply Gaussian blur to reduce noise and improve edge detection
imgBlur = cv2.GaussianBlur(imgGray, (7, 7), 1)

# Detect edges using Canny edge detector
imgCanny = cv2.Canny(imgBlur, 50, 50)

# Make a copy of the original image to draw contours and rectangles
imgContour = img.copy()

# Call the function to find and draw contours
getContours(imgCanny)

# Create a blank image (all zeros) with the same shape as the original
imgBlank = np.zeros_like(img)

# Stack all images for visualization (original, gray, blur, canny, contour, blank)
imgStack = stackImages(0.5, ([img, imgGray, imgBlur],
                             [imgCanny, imgContour, imgBlank]))

# Show the stacked output in a window
cv2.imshow('stacked output', imgStack)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

##### Face detection with viola & jones 

```python
# Import OpenCV library
import cv2

# Load the pre-trained face detection model (Haar Cascade classifier)
# This XML file contains the trained data for detecting faces
faceCascade = cv2.CascadeClassifier("resources/haarcascade_frontalface_default.xml")

# Read the image where we want to detect faces
img = cv2.imread('resources/lena.png')

# Convert the image to grayscale
# Face detection works better on grayscale images because it reduces complexity
imgGray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Detect faces in the grayscale image
# detectMultiScale returns a list of rectangles (x, y, w, h) for each detected face
# Parameters:
#   imgGray: input image
#   1.1: scaleFactor (how much the image size is reduced at each image scale)
#   4: minNeighbors (how many neighbors each candidate rectangle should have to retain it)
faces = faceCascade.detectMultiScale(imgGray, 1.1, 4)

# Loop through all detected faces and draw rectangles around them
for (x, y, w, h) in faces:
    # Draw a rectangle on the original image
    # Parameters:
    #   img: image to draw on
    #   (x, y): top-left corner of rectangle
    #   (x+w, y+h): bottom-right corner of rectangle
    #   (255, 0, 0): color (blue in BGR)
    #   2: thickness of rectangle border
    cv2.rectangle(img, (x, y), (x+w, y+h), (255, 0, 0), 2)

# Show the result image with rectangles drawn
cv2.imshow('output', img)

# Wait for a key press to close the window
cv2.waitKey(0)

# Destroy all OpenCV windows
cv2.destroyAllWindows()
```
